```python
import math
import openpyxl

def generate_minor_project_report(
    input_file="12627223.xlsx",
    output_file="CAP776_Minor_Project_Report_12627223.xlsx",
):
    # Step 1: Open existing workbook (Classroom Deck 8 & 9)
    wb = openpyxl.load_workbook(input_file, data_only=True)
    ws_log = wb["Daily Log"]

    # Read student details
    student_name = ws_log["B2"].value or "Manish Kumar"
    reg_no = ws_log["F2"].value or 12627223
    course_sec = ws_log["B3"].value or "CAP776 / D1P2632"

    # Score lookup dictionaries
    feeling_scores = {
        "Excellent": 5,
        "Good": 4,
        "Neutral": 3,
        "Low": 2,
        "Stressed": 1,
    }
    satisfaction_scores = {
        "Very Satisfied": 5,
        "Satisfied": 4,
        "Neutral": 3,
        "Unsatisfied": 2,
        "Very Unsatisfied": 1,
    }
    energy_scores = {"High": 3, "Medium": 2, "Low": 1}

    # Data collectors
    sleep_list = []
    fitness_list = []
    study_list = []
    coding_list = []
    class_list = []
    other_list = []
    tracked_list = []
    free_list = []
    ei_list = []
    dates_list = []

    # Step 2: Read data rows using loop (Classroom Deck 10 & 11)
    for row in range(6, ws_log.max_row + 1):
        dt_val = ws_log.cell(row=row, column=1).value
        if dt_val is None:
            continue

        dates_list.append(dt_val)
        sleep_list.append(float(ws_log.cell(row=row, column=2).value or 0))
        fitness_list.append(float(ws_log.cell(row=row, column=3).value or 0))
        study_list.append(float(ws_log.cell(row=row, column=4).value or 0))
        coding_list.append(float(ws_log.cell(row=row, column=5).value or 0))
        class_list.append(float(ws_log.cell(row=row, column=6).value or 0))
        other_list.append(float(ws_log.cell(row=row, column=8).value or 0))
        tracked_list.append(float(ws_log.cell(row=row, column=9).value or 0))
        free_list.append(float(ws_log.cell(row=row, column=10).value or 0))

        feel_val = ws_log.cell(row=row, column=11).value
        sat_val = ws_log.cell(row=row, column=12).value
        eng_val = ws_log.cell(row=row, column=13).value

        f_score = feeling_scores.get(str(feel_val).strip(), 3)
        s_score = satisfaction_scores.get(str(sat_val).strip(), 3)
        e_score = energy_scores.get(str(eng_val).strip(), 2)

        daily_ei = (f_score + s_score + e_score) / 3.0
        ei_list.append(daily_ei)

    # Step 3: Compute averages and metrics (Classroom Deck 17)
    valid_days = len(dates_list)
    min_date = min(dates_list)
    max_date = max(dates_list)
    expected_days = (
        (max_date - min_date).days + 1 if hasattr(min_date, "days") else 40
    )
    if not isinstance(expected_days, int):
        expected_days = (max_date - min_date).days + 1
    missing_days = expected_days - valid_days
    invalid_records = 0

    avg_sleep = sum(sleep_list) / valid_days
    avg_fitness = sum(fitness_list) / valid_days
    avg_study = sum(study_list) / valid_days
    avg_coding = sum(coding_list) / valid_days
    avg_class = sum(class_list) / valid_days
    avg_other = sum(other_list) / valid_days
    avg_tracked = sum(tracked_list) / valid_days
    avg_free = sum(free_list) / valid_days

    # Index values
    tpi = avg_coding
    aai = avg_study + avg_class
    phai = avg_fitness
    sri = avg_sleep
    abi = avg_free
    tui = avg_tracked
    ei = sum(ei_list) / valid_days
    dci = (valid_days / expected_days) * 100.0

    pai = (
        0.15 * tpi
        + 0.20 * aai
        + 0.15 * phai
        + 0.20 * sri
        + 0.15 * tui
        + 0.10 * ei
        + 0.05 * dci
    )

    # Step 4: Create new workbook and write structured tables (Classroom Deck 4, 5, 6 & 15)
    out_wb = openpyxl.Workbook()
    ws_rep = out_wb.active
    ws_rep.title = "Intelligence Report"
    ws_rep.views.sheetView[0].showGridLines = True

    # Styling components
    title_font = Font(name="Calibri", size=14, bold=True, color="1F4E78")
    header_font = Font(name="Calibri", size=11, bold=True, color="FFFFFF")
    section_font = Font(name="Calibri", size=12, bold=True, color="1F4E78")
    bold_font = Font(name="Calibri", size=11, bold=True)
    regular_font = Font(name="Calibri", size=11)

   

    # Report Title
    ws_rep["A1"] = (
        "CAP776: PROGRAMMING IN PYTHON — MINOR PROJECT #1 INTELLIGENCE REPORT"
    )
    ws_rep["A1"].font = title_font
    ws_rep.merge_cells("A1:D1")

    # Table 0: Student Details
    ws_rep.append([])
    ws_rep.append(["Student Details", "Info"])
    for col in range(1, 3):
        cell = ws_rep.cell(row=ws_rep.max_row, column=col)
        cell.font = header_font
        cell.fill = navy_fill

    date_str_start = (
        min_date.strftime("%d %B %Y")
        if hasattr(min_date, "strftime")
        else str(min_date)
    )
    date_str_end = (
        max_date.strftime("%d %B %Y")
        if hasattr(max_date, "strftime")
        else str(max_date)
    )

    student_details = [
        ["Student Name", student_name],
        ["Registration / Roll No.", reg_no],
        ["Program / Section", course_sec],
        ["Activity Recording Period", f"{date_str_start} to {date_str_end}"],
        ["Date of Report Generation", f"{date_str_end}"],
    ]
    for row_item in student_details:
        ws_rep.append(row_item)
        for col in range(1, 3):
            cell = ws_rep.cell(row=ws_rep.max_row, column=col)
            cell.font = regular_font
            cell.border = thin_border

    # Table 1: Activity Data Summary
    ws_rep.append([])
    ws_rep.append(["1. Activity Data Summary", ""])
    ws_rep.cell(row=ws_rep.max_row, column=1).font = section_font

    ws_rep.append(["Particular", "Value"])
    for col in range(1, 3):
        cell = ws_rep.cell(row=ws_rep.max_row, column=col)
        cell.font = header_font
        cell.fill = steel_fill

    summary_rows = [
        ["Expected number of days", f"{expected_days} days"],
        ["Valid days recorded", f"{valid_days} days"],
        ["Missing days", f"{missing_days} days"],
        ["Invalid / excluded records", f"{invalid_records} records"],
        ["Average Sleep/day", f"{avg_sleep:.2f} min/day"],
        ["Average Fitness/day", f"{avg_fitness:.2f} min/day"],
        ["Average Study/day", f"{avg_study:.2f} min/day"],
        ["Average Coding/day", f"{avg_coding:.2f} min/day"],
        ["Average Class/day", f"{avg_class:.2f} min/day"],
        ["Average Other Activities/day", f"{avg_other:.2f} min/day"],
        ["Average Free / Unaccounted Time/day", f"{avg_free:.2f} min/day"],
    ]
    for row_item in summary_rows:
        ws_rep.append(row_item)
        for col in range(1, 3):
            cell = ws_rep.cell(row=ws_rep.max_row, column=col)
            cell.font = regular_font
            cell.border = thin_border

    # Table 2: Index Values
    ws_rep.append([])
    ws_rep.append(["2. Index Values", "", "", ""])
    ws_rep.cell(row=ws_rep.max_row, column=1).font = section_font

    ws_rep.append(["Index Name", "Acronym", "Formula", "Value to Enter"])
    for col in range(1, 5):
        cell = ws_rep.cell(row=ws_rep.max_row, column=col)
        cell.font = header_font
        cell.fill = steel_fill

    index_table = [
        [
            "Personal Activity Index",
            "PAI",
            (
                "PAI = 0.15(TPI) + 0.20(AAI) + 0.15(PhAI) + 0.20(SRI) +"
                " 0.15(TUI) + 0.10(EI) + 0.05(DCI)"
            ),
            f"{pai:.2f}",
        ],
        [
            "Tech Productivity",
            "TPI",
            "TPI = Average(Coding Time per valid recorded day)",
            f"{tpi:.2f} min/day",
        ],
        [
            "Academic Activity",
            "AAI",
            "AAI = Average(Study Time + Class Time per valid recorded day)",
            f"{aai:.2f} min/day",
        ],
        [
            "Physical Activity",
            "PhAI",
            "PhAI = Average(Fitness Time per valid recorded day)",
            f"{phai:.2f} min/day",
        ],
        [
            "Sleep & Recovery",
            "SRI",
            "SRI = Average(Sleep Time per valid recorded day)",
            f"{sri:.2f} min/day",
        ],
        [
            "Activity Balance",
            "ABI",
            "ABI = Average(Free / Unaccounted Time per valid recorded day)",
            f"{abi:.2f} min/day",
        ],
        [
            "Time Utilization",
            "TUI",
            "TUI = Average(Total Tracked Time per valid recorded day)",
            f"{tui:.2f} min/day",
        ],
        [
            "Experience Index",
            "EI",
            (
                "EI = Average of daily (Feeling Score + Satisfaction Score +"
                " Energy Score) / 3"
            ),
            f"{ei:.2f} / 5",
        ],
        [
            "Data Continuity Index",
            "DCI",
            "DCI = (Valid Recorded Days / Expected Days) * 100",
            f"{dci:.2f} %",
        ],
    ]

    for idx, row_item in enumerate(index_table):
        ws_rep.append(row_item)
        for col in range(1, 5):
            cell = ws_rep.cell(row=ws_rep.max_row, column=col)
            cell.border = thin_border
            if idx == 0:
                cell.font = bold_font
                cell.fill = highlight_fill
            else:
                cell.font = regular_font

    # Step 5: Add Bar Chart (Classroom Deck 18)
    chart_start_row = ws_rep.max_row + 2
    ws_rep.cell(row=chart_start_row, column=1, value="Activity Indicator")
    ws_rep.cell(row=chart_start_row, column=2, value="Minutes / Day")
    for col in range(1, 3):
        cell = ws_rep.cell(row=chart_start_row, column=col)
        cell.font = header_font
        cell.fill = navy_fill

    chart_items = [
        ("Tech Productivity (TPI)", round(tpi, 2)),
        ("Academic Activity (AAI)", round(aai, 2)),
        ("Physical Activity (PhAI)", round(phai, 2)),
        ("Sleep & Recovery (SRI)", round(sri, 2)),
        ("Free Time (ABI)", round(abi, 2)),
    ]
    for offset, (lbl, val) in enumerate(chart_items, start=1):
        r = chart_start_row + offset
        ws_rep.cell(row=r, column=1, value=lbl)
        ws_rep.cell(row=r, column=2, value=val)

    chart = BarChart()
    chart.type = "col"
    chart.title = "Activity Time Allocation"
    chart.y_axis.title = "Minutes"
    chart.x_axis.title = "Category"
    chart.width = 16
    chart.height = 9

    chart_data = Reference(
        ws_rep,
        min_col=2,
        min_row=chart_start_row,
        max_row=chart_start_row + len(chart_items),
    )
    chart_cats = Reference(
        ws_rep,
        min_col=1,
        min_row=chart_start_row + 1,
        max_row=chart_start_row + len(chart_items),
    )
    chart.add_data(chart_data, titles_from_data=True)
    chart.set_categories(chart_cats)
    chart.legend = None
    ws_rep.add_chart(chart, f"D{chart_start_row}")

    # Column dimensions (Classroom Deck 15)
    col_widths = {"A": 36, "B": 14, "C": 70, "D": 22}
    for col_letter, width in col_widths.items():
        ws_rep.column_dimensions[col_letter].width = width

    # Step 6: Save workbook (Classroom Deck 7 & 8)
    out_wb.save(output_file)
    print("Report generated and saved to:", output_file)


if __name__ == "__main__":
    generate_minor_project_report()
```
